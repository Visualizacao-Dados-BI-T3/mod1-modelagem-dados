Leia 2 datas (cada data é composta por 3 variáveis inteiras: dia, mês e ano) e escreva qual delas é a mais recente.

1.   Item da lista
2.   Item da lista




```python
# ENTRADA = 2 DATAS, 3 VARIAVEIS PRA CADA DATA
print("---- PRIMEIRA DATA ----")
dia1 = int(input("Dia: " ))
mes1 = int(input("Mês: " ))
ano1 = int(input("Ano: " ))

print("---- SEGUNDA DATA ----")
dia2 = int(input("Dia: " ))
mes2 = int(input("Mês: " ))
ano2 = int(input("Ano: " ))

if ano1 > ano2:  # 2026 > 2025
  print("A primeira data é mais recente")

if ano2 > ano1: #  2025 < 2026
  print("A segunda data é mais recente")

if ano1 == ano2: # 2026
  if mes1 > mes2: # 08-2026 > 07-2026
    print("A primeira data é mais recente")
  elif mes2 > mes1: # # 06-2026 < 07-2026
    print("A segunda data é mais recente")
  else: # significa que sao iguais
    if dia1 > dia2: # 03-08-2026 >  02-08-2026
      print("A primeira data é mais recente")
    elif dia2 > dia1: # 01-08-2026 < 02-08-2026
      print("A segunda data é mais recente")
    else:
      print("A data é igual!")
```

* o salários até R$ 280,00 (incluindo) : aumento de 20%

* o salários entre R$ 280,00 e R$ 700,00 : aumento de 15%
* o salários entre R$ 700,00 e R$ 1500,00 : aumento de 10%
* o salários de R$ 1500,00 em diante : aumento de 5%
Após o aumento ser realizado, informe na tela:
* o salário antes do reajuste;
* o percentual de aumento aplicado;
* o valor do aumento;
* o novo salário, após o aumento.


```python
salario = float(input("Digite o salario: ")) # 280
reajuste = int(input("Digite o reajuste: "))

# ordem de precedencia -

if salario <= 280:
  aumento = (20 / 100 * salario)
  salario_novo = salario + aumento # salario antigo + porcentagem do salario
  print(f"Salario antigo {salario}, percentual: 20%, aumento: {aumento}, salario novo: {salario_novo}")

elif salario > 280 and salario <= 700:
  salario_novo = salario + (15 / 100 * salario) # salario antigo + porcentagem do salario
  print(f"Salario antigo {salario}, percentual: 15%, salario novo: {salario_novo}")

elif salario > 700 and salario <= 1500:
  salario_novo = salario + (10 / 100 * salario)
  print(f"Salario antigo {salario}, percentual: 10%, salario novo: {salario_novo}")

elif salario > 1500:  # poderia ser else, já que todas as outras ja foram testadas
  salario_novo = salario + (5 / 100 * salario)
  print(f"Salario antigo {salario}, percentual: 5%, salario novo: {salario_novo}")

```
