# 1. Leia uma velocidade em m/s, calcule e escreva esta velocidade em km/h. (Vkm/h = Vm/s * 3.6)


```python
# Questão 1

# recebendo do usuário
velocidade_m_s = float(input("Digite a velocidade em m/s: "))

# processamento
velocidade_km_h = velocidade_m_s * 3.6

# output
print("A velocidade é ", velocidade_km_h)
# print("A velocidade é " + velocidade_km_h) # erro, porque não posso somar string com num
print(f"A velocidade em km/h é: {velocidade_km_h:.3f}")
```

# 3. Leia um valor em minutos, calcule e escreva o equivalente em horas e minutos.


```python
# Questão 3

minutos = int(input("Minutos: "))

# 60 min = 1h, logo, preciso dividir por 60 pra ter horas exatas
horas =  minutos // 60

minutos_restantes = minutos % 60

print("Em", minutos, "minutos há", horas, "horas e", minutos_restantes, "minutos")
print(f"Em {minutos} minutos há {horas} horas e {minutos_restantes} minutos")
```

# 5. Leia um número inteiro (3 dígitos), calcule e escreva a soma de seus elementos (C + D + U).

```python
num = int(input("Digite um número de três dígitos: ")) #150

# como pegar a dezena?
centena = num // 100 # parte inteira da divisão por 100, por ex: 150 // 100 fica 1

dezena = (num - (centena * 100)) // 10

unidade = num % 10 # se eu pego o resto da divisão por dez, sobra a unidade. por ex: 150 % 10 = 150 / 10 = 15 e resto 0

print("{centena} + {dezena} + unidade")
```

# 7. Leia 3 números, calcule e escreva a soma dos 2 primeiros e a diferença entre os 2 últimos.


```python
num1 = int(input("Digite o primeiro número: "))
num2 = int(input("Digite o segundo número: "))
num3 = int(input("Digite o terceiro número: "))

soma_dois_primeiros = num1 + num2
diferenca_dois_ultimos = num2 - num3

print(f"{num1} + {num2} = {soma_dois_primeiros}")
print(f"{num2} - {num3} = {diferenca_dois_ultimos}")
```

# 12. Leia o salário de um trabalhador e escreva seu novo salário com um aumento de 25%.

```python
# Questão 12

print("Digite seu salário: ")

salario = input()

salario_float = float(salario)

porcentagem_acrescimo = 25 / 100

acrescimo = salario_float * porcentagem_acrescimo

total_recebido = salario_float + acrescimo

print("Acréscimo: ", acrescimo)
print(f"Salário recebido: {total_recebido}")
```
