21. Leia 2 datas (cada data é composta por 3 variáveis inteiras: dia, mês e ano) e escreva qual delas é a mais recente.

Opção 1

```python
# ENTRADA = 2 DATAS, 3 VARIAVEIS PRA CADA DATA
print("---- PRIMEIRA DATA ----")
dia1 = int(input("Dia: " ))
mes1 = int(input("Mês: " ))
ano1 = int(input("Ano: " ))

print("---- SEGUNDA DATA ----")
dia2 = int(input("Dia: " ))
mes2 = int(input("Mês: " ))
ano2 = int(input("Ano: " ))

if ano1 > ano2:  # 2026 > 2025
  print("A primeira data é mais recente")

if ano2 > ano1: #  2025 < 2026
  print("A segunda data é mais recente")

if ano1 == ano2: # 2026
  if mes1 > mes2: # 08-2026 > 07-2026
    print("A primeira data é mais recente")
  elif mes2 > mes1: # # 06-2026 < 07-2026
    print("A segunda data é mais recente")
  else: # significa que sao iguais
    if dia1 > dia2: # 03-08-2026 >  02-08-2026
      print("A primeira data é mais recente")
    elif dia2 > dia1: # 01-08-2026 < 02-08-2026
      print("A segunda data é mais recente")
    else:
      print("A data é igual!")
```

Leia 2 (dois) números, verifique e escreva o menor e o maior entre os números lidos.


```python
# ENTRADA -> DOIS NUMEROS QUE O USUARIO VAI DIGITAR
numero1 = float(input("Digite o primeiro número: "))
numero2 = float(input("Digite o segundo número: "))

# PROCESSAMENTO - COMPARAR OS DOIS >, <, >=,<=, !=, ==
if numero1 > numero2:
  # SAÍDA - DIZER QUAL O MAIOR E O MENOR
  print(f"{numero1} > {numero2}")
elif numero1 == numero2:
  print(f"Numero {numero1} igual a {numero2}")
else:
  print(f"Numero {numero2} é maior e {numero1} é menor")



```
