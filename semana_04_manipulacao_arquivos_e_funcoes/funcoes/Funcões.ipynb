```python
def maior_e_menor(*agrs):
  menor = min(agrs)
  maior = max(agrs)

  return maior, menor

numeros = [2,5,-9,9,87,0]

maior, menor = maior_e_menor(*numeros)

print(maior)
print(menor)
```
