```python
import pandas as pd

# Dicionário onde cada chave é uma coluna
lista_produtos = {
    "Produto": ["Notebook", "Celular"],
    "Preco": [1520, 15555]
}

# Criando o DataFrame a partir do dicionário
df_dict = pd.DataFrame(lista_produtos)

print("--- DataFrame vindo de Dicionário ---")
print(df_dict)

# Acessando apenas uma coluna
print("\nColuna de Preços:")
print(df_dict["Preco"])
```

```python
import pandas as pd

# Matriz (apenas os dados das linhas)
matriz = [
    ["Notebook", 1520],
    ["Celular", 15555]
]

# Criando o DataFrame definindo as colunas no parâmetro 'columns'
df_matriz = pd.DataFrame(matriz, columns=["Produto", "Preco"])

print("--- DataFrame vindo de Matriz ---")
print(df_matriz)
```

```python
# Exportar todas as colunas
df_dict.to_excel("produtos_final.xlsx", index=False, engine="openpyxl")

# Ou exportar apenas colunas específicas
df_dict.to_excel("produtos_apenas_nome.xlsx", index=False, columns=["Produto"], engine="openpyxl")
```
