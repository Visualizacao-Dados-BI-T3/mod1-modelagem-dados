```python
lista_numeros = [1,2,5,8,3,5,6,7]

# ordenar in place
lista_numeros.sort()
print(lista_numeros)

# ordenar em novo objeto
lista_nova = sorted(lista_numeros)
print(f"Lista nova: {lista_nova}")
print(f"Lista antiga: {lista_numeros}")

# inverter in place (na propria lista)
lista_numeros.reverse()
print(lista_numeros)

# inverter em um novo objeto
lista_nova = list(reversed(lista_numeros)) # para nao retornar o tipo de objeto, preciso fazer casting
print(f"Lista nova: {lista_nova}")
print(f"Lista antiga: {lista_numeros}")
```

### Questão 4: Leia 2 arrays A e B com N elementos, escreva um array C, que represente o conjunto união entre os arrays A e B; e um array D, que represente o conjunto interseção entre os arrays A e B.

```python
A = [1,2,2,3,9]
B = [7,89,10,2]

C = set(A).union(B)
D = set(A).intersection(B)

print(C)
print(D)
```
